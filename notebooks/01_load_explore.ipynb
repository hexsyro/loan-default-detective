## 📓 Notebook 01: Load & Explore

**Skills practiced:** `.info()`, checking target class balance with `value_counts(normalize=True)`.

**Goal:** Load the lending dataset and record the exact default/no-default class balance — this number matters later when judging accuracy.

```python
import pandas as pd

path='../data/raw/raw_loan_data.csv'
cols = ['title', 'loan_amnt', 'int_rate', 'dti', 'revol_util','open_acc','installment','emp_title', 'emp_length', 'home_ownership', 'annual_inc', 'loan_status']
df=pd.read_csv(path, usecols=cols)
df=df.rename(columns={'title':'purpose'})
df.info()
df.head()
```

```python

df.describe()
```

```python
df.groupby('loan_status')['emp_title'].count()
```

```python
df = df[df["loan_status"].isin([
    "Fully Paid",
    "Charged Off",
    "Does not meet the credit policy. Status:Charged Off",
    "Default"
])]
```

```python
df['defaulted']=(df['loan_status'].isin(['Default',
                                        'Does not meet the credit policy. Status:Charged Off',
                                        'Charged Off'])).astype(int)
```

```python
print(df['defaulted'].value_counts(normalize=True))
df.to_csv('../data/processed/loans_cleaned.csv', index=False)
```
